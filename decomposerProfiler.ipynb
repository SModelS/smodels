```python
import cProfile
import pstats
import io


def _profile_decomposition(fn, label, model, sigmacut, clear_cache=True, top_n=20, **kwargs):
    """Profile fn(model, sigmacut, **kwargs), optionally clearing particle caches first."""
    if clear_cache:
        clear_particle_caches(model)

    profiler = cProfile.Profile()
    t0 = time.perf_counter()
    profiler.enable()
    result = fn(model, sigmacut, **kwargs)
    profiler.disable()
    dt = time.perf_counter() - t0

    stats_stream = io.StringIO()
    pstats.Stats(profiler, stream=stats_stream).sort_stats("cumulative").print_stats(top_n)

    print(f"\n=== {label} — {dt:.3f} s ===")
    print(f"n_topologies: {len(result)},  n_sms: {len(result.getSMSList())}")
    print(stats_stream.getvalue())

    return result, profiler, dt


_kw = dict(massCompress=massCompress, invisibleCompress=invisibleCompress,
           minmassgap=mingap, minmassgapISR=mingapISR)

# Each call clears particle caches so every run is an independent cold start
res_old, prof_old, dt_old = _profile_decomposition(decomposeOld, "Old (BFS)",           model, sigmacut, **_kw)
res_cur, prof_cur, dt_cur = _profile_decomposition(decompose,    "Current (DFS)",       model, sigmacut, **_kw)
res_new, prof_new, dt_new = _profile_decomposition(decomposeNew, "New (subtree cache)", model, sigmacut, **_kw)

print("\n── Wall-clock summary ────────────────────────────────────────")
print(f"  Old (BFS):              {dt_old:.3f} s  (baseline)")
print(f"  Current (DFS):          {dt_cur:.3f} s  ({dt_old/dt_cur:.2f}× vs Old)")
print(f"  New (subtree cache):    {dt_new:.3f} s  ({dt_old/dt_new:.2f}× vs Old)")
```

```python
def _cumtime_for(stats_obj, needle):
    """Sum cumulative seconds for all profiled functions whose 'file:name' contains needle."""
    total = 0.0
    for (filename, _lineno, funcname), stat in stats_obj.stats.items():
        if needle in f"{filename}:{funcname}":
            total += stat[3]   # cumtime is index 3
    return total


stats_old = pstats.Stats(prof_old)
stats_cur = pstats.Stats(prof_cur)
stats_new = pstats.Stats(prof_new)

# (display label, list of substrings matched against "file:funcname")
hotspots = [
    ("Cascade / subtree build",  ["iterCascadeDecay", "addOneStepDecays", "build_subtree_cache"]),
    ("compress (total)",         ["theorySMS.py:compress"]),
    ("  massCompress",           ["theorySMS.py:massCompress"]),
    ("  invisibleCompress",      ["theorySMS.py:invisibleCompress"]),
    ("addSMS",                   ["topologyDict.py:addSMS"]),
    ("sort / compareSubTrees",   ["genericSMS.py:sort", "genericSMS.py:compareSubTrees"]),
    ("setGlobalProperties",      ["theorySMS.py:setGlobalProperties"]),
    ("copy",                     ["theorySMS.py:copy"]),
    ("getDecayNodes",            ["decomposer.py:getDecayNodes", "decomposerOld.py:getDecayNodes"]),
]

COL = 14
hdr = f"{'Hotspot':<36} {'Old (BFS)':>{COL}} {'Current (DFS)':>{COL}} {'New (cache)':>{COL}}"
sep = "─" * len(hdr)
print(hdr)
print(sep)
for lbl, needles in hotspots:
    t_old = sum(_cumtime_for(stats_old, n) for n in needles)
    t_cur = sum(_cumtime_for(stats_cur, n) for n in needles)
    t_new = sum(_cumtime_for(stats_new, n) for n in needles)
    print(f"{lbl:<36} {t_old:>{COL}.3f} {t_cur:>{COL}.3f} {t_new:>{COL}.3f}")

print(sep)
print(f"{'TOTAL':>36} {dt_old:>{COL}.3f} {dt_cur:>{COL}.3f} {dt_new:>{COL}.3f}")
```
